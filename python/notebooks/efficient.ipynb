```python
# ============================================================
#  BASELINE TRAINING CELL — corrected for fair comparison
#  ... [Your original comments preserved] ...
# ============================================================

from google.colab import drive
from pathlib import Path
import sys, os, json, random, copy
from datetime import datetime, timezone

import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
from torch.utils.data import Dataset

# ── 1. Mount Drive, Transfer to Local, and set project root ──
drive.mount('/content/drive', force_remount=False)

DRIVE_ROOT = Path('/content/drive/Othercomputers/My Laptop/thesis_project')
LOCAL_ROOT = Path('/content/thesis_project')

print("Copying baseline datasets and code to local fast storage... (This will take a moment)")
!mkdir -p "/content/thesis_project"
# Using -u (update) so it doesn't redundantly copy files if you run the cell twice
!cp -r -u "{DRIVE_ROOT}/python" "/content/thesis_project/"
!cp -r -u "{DRIVE_ROOT}/artifacts" "/content/thesis_project/"

# Set the active root to the lightning-fast local Colab disk
PROJECT_ROOT = LOCAL_ROOT

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

os.chdir(PROJECT_ROOT)
print("Working directory now local:", os.getcwd())

# ── 2. Imports (after path is set) ───────────────────────────
from python.src.legacy_models import (
    LiteratureBaseline_VGG16,
    LiteratureBaseline_ResNet18,
    LiteratureBaseline_DenseNet121,
)
from python.src.dataio import load_artifact
from python.src.preprocessing import build_feature_tensor, split_dataset
from python.src.utils import (
    create_train_loader, create_eval_loader,
    resolve_device, FeatureTensorDataset,
)

# ── 3. Config ─────────────────────────────────────────────────
SEEDS         = [38, 55, 73]    # multi-seed for proper mean ± std
N_PER_CLASS   = 2500
SPEC_VER      = "v2"

PHASE1_EPOCHS = 5               # head-only warm-up (linear head saturates fast)
PHASE2_MAX_EP = 30              # cap; early stopping usually triggers earlier
LR_PHASE1     = 1e-3            # Adam, head only
LR_PHASE2     = 1e-3            # SGD+momentum on last block
SGD_MOMENTUM  = 0.9
WEIGHT_DECAY  = 5e-4            # standard fine-tuning WD with SGD
BATCH_SIZE    = 32
NUM_CLASSES   = 10
EARLY_STOP_PATIENCE = 8         # epochs of no val_loss improvement → stop

# SpecAugment parameters (applied on normalised STFT, train batches only)
SPEC_AUG_FREQ_MASKS = 1
SPEC_AUG_TIME_MASKS = 1
SPEC_AUG_FREQ_FRAC  = 0.10      # mask up to 10% of the frequency axis
SPEC_AUG_TIME_FRAC  = 0.10      # mask up to 10% of the time axis

BASELINE_REGISTRY = {
    "vgg_16":       LiteratureBaseline_VGG16,
    "resnet_18":    LiteratureBaseline_ResNet18,
    "densenet_121": LiteratureBaseline_DenseNet121,
}

# ── 4. Helpers ────────────────────────────────────────────────

def _set_seed(seed):
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    np.random.seed(seed)
    random.seed(seed)
    torch.backends.cudnn.deterministic = True


def _compute_stft_stats(x_stft_train: torch.Tensor):
    mean = x_stft_train.mean(dim=(0, 2, 3), keepdim=True)
    std  = x_stft_train.std (dim=(0, 2, 3), keepdim=True).clamp_min(1e-6)
    return mean, std


def _normalise(x_stft, mean, std):
    return (x_stft - mean) / std


def _spec_augment(x_stft: torch.Tensor) -> torch.Tensor:
    if not torch.is_grad_enabled():   # safety: only augment in training
        return x_stft
    B, C, F_dim, T_dim = x_stft.shape
    out = x_stft.clone()
    max_f = max(1, int(F_dim * SPEC_AUG_FREQ_FRAC))
    max_t = max(1, int(T_dim * SPEC_AUG_TIME_FRAC))
    for b in range(B):
        for _ in range(SPEC_AUG_FREQ_MASKS):
            w = torch.randint(0, max_f + 1, (1,)).item()
            if w > 0:
                f0 = torch.randint(0, F_dim - w + 1, (1,)).item()
                out[b, :, f0:f0 + w, :] = 0.0
        for _ in range(SPEC_AUG_TIME_MASKS):
            w = torch.randint(0, max_t + 1, (1,)).item()
            if w > 0:
                t0 = torch.randint(0, T_dim - w + 1, (1,)).item()
                out[b, :, :, t0:t0 + w] = 0.0
    return out


def _make_loader(x_stft, x_iq, x_if, y, batch_size, shuffle):
    ds = FeatureTensorDataset(x_stft, x_iq, x_if, y)
    return create_train_loader(ds, batch_size) if shuffle else create_eval_loader(ds, batch_size)


@torch.no_grad()
def _evaluate(model, loader, criterion, device):
    model.eval()
    total_loss, total_correct, total_n = 0.0, 0, 0
    for x_stft, x_iq, x_if, y in loader:
        x_stft, x_iq, x_if, y = x_stft.to(device), x_iq.to(device), x_if.to(device), y.to(device)
        logits = model(x_stft, x_iq, x_if)
        total_loss    += criterion(logits, y).item() * y.size(0)
        total_correct += (logits.argmax(1) == y).sum().item()
        total_n       += y.size(0)
    return total_loss / total_n, total_correct / total_n


def _run_phase(model, train_loader, val_loader, optimizer, scheduler,
               criterion, device, start_ep, max_ep, phase_label,
               log_epochs, augment, use_plateau_sched,
               best_val_loss_init=float("inf"),
               patience=EARLY_STOP_PATIENCE):
    best_acc, best_state = 0.0, None
    best_val_loss = best_val_loss_init
    epochs_no_improve = 0
    last_epoch = start_ep - 1

    for epoch in range(start_ep, max_ep + 1):
        model.train()
        total_loss, total_correct, total_n = 0.0, 0, 0
        for x_stft, x_iq, x_if, y in train_loader:
            x_stft = x_stft.to(device)
            if augment:
                x_stft = _spec_augment(x_stft)
            x_iq, x_if, y = x_iq.to(device), x_if.to(device), y.to(device)

            optimizer.zero_grad()
            logits = model(x_stft, x_iq, x_if)
            loss   = criterion(logits, y)
            loss.backward()
            optimizer.step()
            bs = y.size(0)
            total_loss    += loss.item() * bs
            total_correct += (logits.argmax(1) == y).sum().item()
            total_n       += bs

        train_loss = total_loss / total_n
        train_acc  = total_correct / total_n
        val_loss, val_acc = _evaluate(model, val_loader, criterion, device)

        if use_plateau_sched:
            scheduler.step(val_loss)
        else:
            scheduler.step()
        lr_now = optimizer.param_groups[0]["lr"]

        if val_acc > best_acc:
            best_acc = val_acc
            best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}

        if val_loss < best_val_loss - 1e-4:
            best_val_loss = val_loss
            epochs_no_improve = 0
        else:
            epochs_no_improve += 1

        if epoch == start_ep or epoch % 2 == 0 or epoch == max_ep:
            print(f"    [{phase_label}] Ep {epoch:02d}/{max_ep} | "
                  f"LR {lr_now:.2e} | Train {train_loss:.4f} ({100*train_acc:.1f}%) | "
                  f"Val {val_loss:.4f} ({100*val_acc:.2f}%) | NoImp {epochs_no_improve}")

        log_epochs.append({"epoch": epoch, "phase": phase_label,
                           "train_loss": train_loss, "train_acc": train_acc,
                           "val_loss": val_loss, "val_accuracy": val_acc, "lr": lr_now})
        last_epoch = epoch

        if phase_label == "Phase2" and epochs_no_improve >= patience:
            print(f"    [{phase_label}] Early stop at epoch {epoch} (no val_loss improvement for {patience} epochs).")
            break

    return best_acc, best_state, best_val_loss, last_epoch


# ── 5. Multi-seed × multi-model training ──────────────────────
device = resolve_device("auto")
print(f"\nDevice: {device}\n")

for seed in SEEDS:
    print(f"\n{'#'*70}")
    print(f"#  SEED {seed}")
    print(f"{'#'*70}")

    print("\nLoading dataset...")
    _train_path = (PROJECT_ROOT / "artifacts" / "datasets" / "impaired"
                   / f"impaired_dataset_{SPEC_VER}_seed{seed}_n{N_PER_CLASS}_train.mat")
    _eval_path  = (PROJECT_ROOT / "artifacts" / "datasets" / "impaired"
                   / f"impaired_dataset_{SPEC_VER}_seed{seed}_n{N_PER_CLASS}_eval.mat")

    _train_artifact = load_artifact(str(_train_path), load_params=False)
    _x_stft, _x_iq, _x_if, _y = build_feature_tensor(_train_artifact)

    train_set, val_set = split_dataset(_x_stft, _x_iq, _x_if, _y,
                                       train_ratio=0.8, seed=seed)

    x_stft_tr  = train_set.dataset.tensors[0][train_set.indices]
    x_iq_tr    = train_set.dataset.tensors[1][train_set.indices]
    x_if_tr    = train_set.dataset.tensors[2][train_set.indices]
    y_tr       = train_set.dataset.tensors[3][train_set.indices]
    x_stft_val = val_set.dataset.tensors[0][val_set.indices]
    x_iq_val   = val_set.dataset.tensors[1][val_set.indices]
    x_if_val   = val_set.dataset.tensors[2][val_set.indices]
    y_val      = val_set.dataset.tensors[3][val_set.indices]

    _eval_artifact = load_artifact(str(_eval_path), load_params=False)
    x_stft_te, x_iq_te, x_if_te, y_te = build_feature_tensor(_eval_artifact)

    mean, std = _compute_stft_stats(x_stft_tr)
    print(f"  STFT channel means: {mean.flatten().tolist()}")
    print(f"  STFT channel stds : {std.flatten().tolist()}")

    x_stft_tr  = _normalise(x_stft_tr,  mean, std)
    x_stft_val = _normalise(x_stft_val, mean, std)
    x_stft_te  = _normalise(x_stft_te,  mean, std)

    def _dist(y):
        return torch.bincount(y, minlength=NUM_CLASSES).tolist()
    print(f"  Train class dist : {_dist(y_tr)}")
    print(f"  Val   class dist : {_dist(y_val)}")
    print(f"  Test  class dist : {_dist(y_te)}")
    print(f"  Train n={x_stft_tr.shape[0]} | Val n={x_stft_val.shape[0]} | Test n={x_stft_te.shape[0]}")

    for model_name, model_cls in BASELINE_REGISTRY.items():
        ckpt_path = (PROJECT_ROOT / "artifacts" / "checkpoints"
                     / f"{model_name}_baseline_seed{seed}_n{N_PER_CLASS}.pt")

        if ckpt_path.exists():
            print(f"\n[SKIP] {model_name} (seed {seed}) — checkpoint already exists.")
            continue

        print(f"\n{'='*65}")
        print(f"  Training: {model_name.upper()}  (seed {seed})")
        print(f"{'='*65}")

        _set_seed(seed)
        model     = model_cls(num_classes=NUM_CLASSES).to(device)
        criterion = nn.CrossEntropyLoss(label_smoothing=0.1)

        train_loader = _make_loader(x_stft_tr,  x_iq_tr,  x_if_tr,  y_tr,  BATCH_SIZE, shuffle=True)
        val_loader   = _make_loader(x_stft_val, x_iq_val, x_if_val, y_val, BATCH_SIZE, shuffle=False)
        test_loader  = _make_loader(x_stft_te,  x_iq_te,  x_if_te,  y_te,  BATCH_SIZE, shuffle=False)

        log_epochs        = []
        global_best_acc   = 0.0
        global_best_state = None

        model.freeze_for_phase1()
        p1_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
        print(f"  Phase 1 trainable params : {p1_params:,}")

        opt1 = torch.optim.Adam([p for p in model.parameters() if p.requires_grad],
                                 lr=LR_PHASE1, weight_decay=WEIGHT_DECAY)
        sch1 = torch.optim.lr_scheduler.CosineAnnealingLR(opt1, T_max=PHASE1_EPOCHS)

        best1, state1, _, _ = _run_phase(
            model, train_loader, val_loader, opt1, sch1,
            criterion, device, 1, PHASE1_EPOCHS,
            "Phase1", log_epochs, augment=False, use_plateau_sched=False,
        )
        if best1 > global_best_acc:
            global_best_acc, global_best_state = best1, state1

        model.unfreeze_for_phase2()
        p2_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
        print(f"  Phase 2 trainable params : {p2_params:,}")

        opt2 = torch.optim.SGD([p for p in model.parameters() if p.requires_grad],
                                lr=LR_PHASE2, momentum=SGD_MOMENTUM,
                                weight_decay=WEIGHT_DECAY, nesterov=True)
        sch2 = torch.optim.lr_scheduler.ReduceLROnPlateau(
            opt2, mode="min", factor=0.5, patience=3,
        )

        best2, state2, _, last_ep = _run_phase(
            model, train_loader, val_loader, opt2, sch2,
            criterion, device,
            PHASE1_EPOCHS + 1, PHASE1_EPOCHS + PHASE2_MAX_EP,
            "Phase2", log_epochs, augment=True, use_plateau_sched=True,
        )
        if best2 > global_best_acc:
            global_best_acc, global_best_state = best2, state2

        model.load_state_dict(global_best_state)
        test_loss, test_acc = _evaluate(model, test_loader, criterion, device)
        print(f"\n  Best Val Acc : {100*global_best_acc:.2f}%")
        print(f"  Test Acc     : {100*test_acc:.2f}%   (loss {test_loss:.4f})")

        ckpt_path.parent.mkdir(parents=True, exist_ok=True)
        torch.save(global_best_state, ckpt_path)
        print(f"  Checkpoint   : {ckpt_path.name}")

        log_dir = PROJECT_ROOT / "artifacts" / "logs" / "baselines"
        log_dir.mkdir(parents=True, exist_ok=True)
        log_path = log_dir / f"{model_name}_baseline_seed{seed}_n{N_PER_CLASS}.json"
        with log_path.open("w") as f:
            json.dump({"model": model_name, "seed": seed, "n_per_class": N_PER_CLASS,
                       "best_val_acc": global_best_acc, "test_acc": test_acc,
                       "test_loss": test_loss,
                       "stft_norm_mean": mean.flatten().tolist(),
                       "stft_norm_std":  std.flatten().tolist(),
                       "epochs": log_epochs}, f, indent=2)
        print(f"  Log          : {log_path.name}")

        del model
        torch.cuda.empty_cache()

print("\n\n✓ All baselines complete across all seeds.")

# ── 6. Backup Local Files to Drive ────────────────────────────
# =================================================================
print("\nAll training complete! Backing up results to Google Drive...")

# 1. Create the parent folders on your Drive just to be safe
!mkdir -p "{DRIVE_ROOT}/reports"
!mkdir -p "{DRIVE_ROOT}/artifacts"

# 2. Copy all generated figures, checkpoints, and logs back to Drive
# The '-u' flag means "update", so it only copies new files (saves time!)
!cp -r -u "{PROJECT_ROOT}/reports/." "{DRIVE_ROOT}/reports/"
!cp -r -u "{PROJECT_ROOT}/artifacts/." "{DRIVE_ROOT}/artifacts/"

print("Backup complete. Safe to close Colab.")
```
